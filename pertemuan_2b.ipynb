2.	Tampilkan image dalam channel Red-Blue dan Green-Blue saja!

```python
img = cv2.imread("/content/drive/MyDrive/RPL/PCVK/KTM lama.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


# cara Split dan Merge channel RGB
R, G, B = cv2.split(img_rgb)

rb = cv2.merge([R*1, G*0, B*1])   # Red + Blue, Green=0
gb = cv2.merge([R*0, G*1, B*1])   # Green + Blue, Red=0

# cara bitwise
#channel_rb = np.array([1, 0, 1])   # [R,G,B]
#rb = img_rgb * channel_rb
#channel_gb = np.array([0, 1, 1])   # [R,G,B]
#gb = img_rgb * channel_gb

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(rb)
plt.title("Red + Blue")

plt.subplot(1,2,2)
plt.imshow(gb)
plt.title("Green + Blue")
plt.show()
```

5.	Buat 5 kotak berbagai ukuran dan warna yang berbeda dalam satu image

```python
img = cv2.imread("/content/drive/MyDrive/RPL/PCVK/KTM lama.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Salin gambar agar tidak merusak aslinya
img_boxes = img_rgb.copy()

# Ambil ukuran gambar
h, w, _ = img_boxes.shape

# Buat 5 kotak random
for i in range(5):
    # Titik awal (x1,y1) acak
    x1 = np.random.randint(0, w-50)
    y1 = np.random.randint(0, h-50)

    # Ukuran kotak acak (50–150 pixel)
    x2 = x1 + np.random.randint(50, 150)
    y2 = y1 + np.random.randint(50, 150)

    # Warna acak (RGB)
    color = (np.random.randint(0,256),
             np.random.randint(0,256),
             np.random.randint(0,256))

    # Gambar rectangle
    cv2.rectangle(img_boxes, (x1,y1), (x2,y2), color, thickness=-1)

# Tampilkan hasil
plt.figure(figsize=(8,6))
plt.imshow(img_boxes)
plt.title("Gambar dengan 5 Kotak Random")
plt.axis("off")
plt.show()
```
