```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://www.greeners.co/wp-content/uploads/2020/11/Perbedaan-Harimau-dan-Macan-3-min.jpg"]

for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range(tinggi):
    image_3[x, int((lebar)/2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tinggi, lebar = image_3.shape[:2]

for y in range(tinggi):
    x = int(round(y * (lebar - 1) / max(1, (tinggi - 1))))
    image_3[y, x] = [255, 255, 255]


for y in range(tinggi):
    x = int(round((lebar - 1) - y * (lebar - 1) / max(1, (tinggi - 1))))
    image_3[y, x] = [255, 255, 255]



final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tinggi, lebar = image_3.shape[:2]

panjang_garis = lebar // 2

cy = tinggi // 2

x1 = (lebar - panjang_garis) // 2
x2 = x1 + panjang_garis

image_3[cy, x1:x2] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tinggi, llebar = image_3.shape[:2]

box_tinggi, box_lebar = tinggi // 4, lebar // 4

cy, cx = tinggi // 2, lebar // 2

y1, y2 = cy - box_tinggi // 2, cy + box_tinggi // 2
x1, x2 = cx - box_lebar // 2, cx + box_lebar // 2

image_3[y1:y2, x1:x2] = [255, 255, 255]


final_frame3 = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame3)
```
